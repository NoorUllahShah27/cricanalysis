```python
# for loop in python

#for loop always working in a sequential manner..

# for <interactive varible> function/condition/varible:
while True:
    num = int(input("Enter a number of table you want to print:"))
    for i in range(1,21):
        print(f" {num} * {i} = {num*i}")
    rerun=int(input("Do you want to continue?: Press any from below: 1 : No, 2 : Yes"))
    if rerun == 1:
        print("Ok thank you")
        break
    elif rerun == 2:
        print("Ok, Continuing...")
        continue
```
