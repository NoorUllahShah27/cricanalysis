## Tax Eligibility Program

```python

name = input("Enter Your Name: ")
salary = eval(input("Enter Your Salary: "))

if salary>=50000:
    print(f"Dear {name} your are tax eligible because your salary {salary} which is >= taxible income and your tax amount is {((salary*12)*0.04)} per anum.")
else:
    print("Your salary is not tax eligible")
```

## Program for Marksheet

```python
name = input("Enter Your Name: ")
sub1 = eval(input("Enter Subject 1 Marks (out of 100): "))
sub2 = eval(input("Enter Subject 2 Marks (out of 100): "))
sub3 = eval(input("Enter Subject 3 Marks (out of 100): "))
sub4 = eval(input("Enter Subject 4 Marks (out of 100): "))
sub5 = eval(input("Enter Subject 5 Marks (out of 100): "))
obtain = sub1+sub2+sub3+sub4+sub5
percentage = (obtain/500)*100

if percentage>=70:
    print(f"Dear {name} you have scored {obtain} marks, and your percentage is {round(percentage,2)}, and your grade is A")
elif percentage>=60:
    print(f"Dear {name} you have scored {obtain} marks, and your percentage is {round(percentage,2)}, and your grade is B")
elif percentage>=50:
    print(f"Dear {name} you have scored {obtain} marks, and your percentage is {round(percentage,2)}, and your grade is C")
elif percentage>=40:
    print(f"Dear {name} you have scored {obtain} marks, and your percentage is {round(percentage,2)}, and your grade is D")
else:
    print(f"Dear {name} you have scored {obtain} marks, and your percentage is {round(percentage,2)}, and your grade is F")
```

## Car Purchasing Decision as per Budget

```python
name = input("Enter Your Name: ")
budget = eval(input("Enter Your Budget: "))
if budget>=1000000:
    print(f"Dear {name} your budget is {budget} and you can buy Car of A categary")
elif budget>=800000:
    print(f"Dear {name} your budget is {budget} and you can buy Car of B categary")
elif budget>=600000:
    print(f"Dear {name} your budget is {budget} and you can buy Car of C categary")
elif budget>=400000:
    print(f"Dear {name} your budget is {budget} and you can buy Car of D categary")
else:
    print(f"Dear {name} your budget is {budget} and you can buy Car of F categary")
```
