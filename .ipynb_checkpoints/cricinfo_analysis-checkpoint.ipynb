```python
df.rename(columns={'Player':'player','Span':'span','Mat':'matches','Inns':'inns','NO':'not_out','Runs':'runs','HS':'high_score','Ave':'avg','BF':'balls_faced','SR':'strike_rate'},inplace=True)
```

```python
df
```

```python
df.columns
```

```python
df.info()
```

```python
df['span']
```

```python
df['span'].str.split(pat="-")
```

```python
df['span'].str.split(pat="-").str[0]
```

```python
df['start_year']=df['span'].str.split(pat="-").str[0].astype(int)
```

```python
df['end_year']=df['span'].str.split(pat="-").str[1].astype(int)
```

```python
df['duration']=df['end_year']-df['start_year']
```

```python
df=df[['player','span','duration','matches','inns','not_out','runs','high_score','avg','balls_faced','strike_rate','100','50','0','4s','6s','start_year','end_year']]
```

```python
df['high_score']=df['high_score'].str.split(pat="*").str[0].astype(int)
```

```python
df['balls_faced']=df['balls_faced'].str.split(pat="+").str[0]
```

```python
df['balls_faced']=df['balls_faced'].replace("-",0).astype(int)
```

```python
df['strike_rate']=df['strike_rate'].replace("-",0).astype(float)
```

```python
df['0']=df['0'].replace("-",0).astype(int)
```

```python
df['country']=df['player'].str.split(pat="(").str[1]
```

```python
df['country']=df['country'].str.split(pat=")").str[0]
```

```python
df
```

```python
df.to_csv("cricinfo_clean_data.csv",index=False)
```
